```python
import sqlite3
import pandas as pd
import json

DB_PATH = "db.sqlite"

conn = sqlite3.connect(DB_PATH)

# Show all tables
tables = pd.read_sql_query("""
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name
""", conn)

display(tables)
```

```python
trials = pd.read_sql_query(
    """
    SELECT *
    FROM trial_data
    ORDER BY id
    """,
    conn
)

if len(trials) > 0:
    payload = trials["payload"].apply(json.loads)
    payload_df = pd.json_normalize(payload)

    results = pd.concat(
        [
            trials.drop(columns=["payload"]).reset_index(drop=True),
            payload_df.reset_index(drop=True)
        ],
        axis=1
    )

    display(results)
else:
    print("No trials saved yet.")
```
